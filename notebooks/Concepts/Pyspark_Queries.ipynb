```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window
from pyspark.sql.functions import col, dense_rank, rank, row_number, lag, lead

spark = SparkSession.builder.appName("EmpDeptWindowFunctions").getOrCreate()

# Employee Table — 10 Records with dept_id
emp_data = [
    (1,  "Alice",   10, "Senior", 95000),
    (2,  "Bob",     20, "Junior", 60000),
    (3,  "Charlie", 10, "Senior", 95000),
    (4,  "Diana",   10, "Lead",  120000),
    (5,  "Eve",     20, "Senior", 80000),
    (6,  "Frank",   30, "Junior", 50000),
    (7,  "Grace",   30, "Lead",  100000),
    (8,  "Henry",   20, "Junior", 50000),
    (9,  "Iris",    40, "Senior", 90000),
    (10, "Jack",    40, "Junior", 65000),
]

# Department Table
dept_data = [
    (10, "Engineering"),
    (20, "Marketing"),
    (30, "Finance"),
    (40, "HR"),
]

df_emp  = spark.createDataFrame(emp_data,  ["emp_id", "name", "dept_id", "level", "salary"])
df_dept = spark.createDataFrame(dept_data, ["dept_id", "dept_name"])

print("===== EMPLOYEE TABLE =====")
df_emp.show()

print("===== DEPARTMENT TABLE =====")
df_dept.show()
```

```python
df_emp.createOrReplaceGlobalTempView("emp")
df_dept.createGlobalTempView("dept")
```

```python
%sql
select Emp_id, name, dept_id, salary, rank() over ( partition by dept_id order by salary desc ) from emp;
```
