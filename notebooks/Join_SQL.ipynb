```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

emp_data = [(1,), (1,), (None,), (None,)]
dept_data = [(1),(2),(2)]

emp_schema  = ["emp_id"]
dept_schema = ["emp_id"]

emp_df  = spark.createDataFrame(emp_data,  emp_schema)
dept_df = spark.createDataFrame(dept_data, dept_schema)

emp_df.createOrReplaceTempView("emp")
dept_df.createOrReplaceTempView("dept")
```

```python
%sql
select  * from emp A INNER  JOIN dept B on A.emp_id = B.emp_id;
```

```python
%sql
select  * from EMP A LEFT JOIN DEPT B on A.emp_id = B.emp_id;
```

```python
%sql
select  * from EMP A RIGHT outer JOIN DEPT B on A.emp_id = B.emp_id;
```

```python
%sql
select  * from EMP A FULL OUTER JOIN DEPT B on A.emp_id = B.emp_id;
```
