```python
#Dict in LIST
races = [
    {"raceId": 1, "year": 2005, "name": "British GP"},
    {"raceId": 2, "year": 2005, "name": "German GP"},
    {"raceId": 3, "year": 2005, "name": "Hungarian GP"}
]

print(races[0]["name"])   # British GP
```

```python
#Dict Nested
races = {
    81: {
        "year": 2005,
        "name": "British Grand Prix",
        "date": "10/7/2005"
    },
    82: {
        "year": 2005,
        "name": "German Grand Prix",
        "date": "24/7/2005"
    }
}

print(races[81]["name"])  
# British Grand Prix
```

```python
#No sql database is just a nested Dictionary 
race = {
    "raceId": 81,
    "year": 2005,
    "circuit": {
        "name": "Silverstone",
        "country": "UK",
        "Visited_cntry": {"UK": 1,"USA": 2, "India": 3}
                }
        }
print (race["year"])                            #level-1 
print (race["circuit"]["country"])              #level-2
print (race["circuit"]["Visited_cntry"]["USA"]) #level-3
```
